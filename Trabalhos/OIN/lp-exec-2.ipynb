# Problem 1

A furniture manufacturer employs 6 skilled and 11 semiskilled workers and produces two products: study table and computer table. A study table requires 2 h of a skilled worker and 2 h of an unskilled worker. A computer table requires 2 h of a skilled worker and 5 h of an unskilled worker. As per the industrial laws, no one is allowed to work for more than 38 h a week. The manufacturer can sell as many tables as he can produce. If the profit for a study table is \$100 and for a computer table \$160, how many study and computer tables should the manufacturer produce in a week in order to maximize the overall profit? Formulate a linear programming model and solve it.

#Análise do Problema 1

Variáveis de decisão:

\( s \) = número de mesas de **estudo** produzidas por semana  
\( c \) = número de mesas de **computador** produzidas por semana  

Função Objetivo
$$
Z = 100s + 160c
$$

Função Objetivo
1. **Mão de obra qualificada (6 × 38h = 228h):**
$$
2s + 2c \leq 228
$$

2. **Mão de obra semiqualificada (11 × 38h = 418h):**
$$
2s + 5c \leq 418
$$

3. **Não negatividade:**
$$
s \geq 0, \quad c \geq 0
$$

```python
# Modelo
model = pyo.ConcreteModel()

# Variáveis de decisão (número de mesas)
model.study = pyo.Var(domain=pyo.NonNegativeReals)     # Study tables
model.computer = pyo.Var(domain=pyo.NonNegativeReals)  # Computer tables

# Função objetivo: maximizar o lucro
model.profit = pyo.Objective(expr=100*model.study + 160*model.computer,
                             sense=pyo.maximize)
```

```python
# Restrições de mão de obra
# Tempo total disponível para trabalhadores qualificados (6 * 38h)
model.skilled_constraint = pyo.Constraint(expr=2*model.study + 2*model.computer <= 6*38)

# Tempo total disponível para trabalhadores semiqualificados (11 * 38h)
model.semiskilled_constraint = pyo.Constraint(expr=2*model.study + 5*model.computer <= 11*38)

# Resolver com GLPK
solver = pyo.SolverFactory("glpk")
result = solver.solve(model, tee=True)

# Exibir resultados
print("Study tables:", model.study.value)
print("Computer tables:", model.computer.value)
print("Lucro máximo: $", pyo.value(model.profit))
```

# Resultados Ótimos
Após a resolução do modelo de Programação Linear, obtemos os seguintes resultados:

- Mesas de **estudo** (Study tables):  
  $$ s \approx 50,67 $$
- Mesas de **computador** (Computer tables):  
  $$ c \approx 63,33 $$
- **Lucro máximo:**  
  $$ Z \approx \$15.200 $$
---

# Interpretação
O resultado mostra que o **gargalo** do processo produtivo está na **mão de obra qualificada** (228h disponíveis).  
O modelo sugere priorizar a produção de **mesas de computador**, que apresentam maior lucro unitário (\$160), mas ainda mantendo parte da produção de **mesas de estudo** para equilibrar o uso da mão de obra semiqualificada.

Como não é possível produzir frações de mesas, na prática o fabricante pode produzir **50 mesas de estudo e 63 mesas de computador**, alcançando praticamente o mesmo lucro ótimo calculado.
